```python
# Decorator to print the operation, inputs and result

def output_decorator(operation):
    def decorator(func):
        def wrapper(*args,**kwargs):
            result=func(*args,**kwargs)# Run the original function
            inputs=", ".join(map(str,args))
            if kwargs:
                extra=", ".join(str(k)+"="+str(v)for k,v in kwargs.items())
                inputs=inputs+", "+extra if inputs else extra

            print("##########################")
            print("#",operation)
            print("# input:",inputs)
            print("# output:",result)
            print("##########################")
            return result
        return wrapper
    return decorator

# Adds two numbers and shows the result in the given format
@output_decorator("Adding")
def add(a,b):
    return a+b

add(4,3)
```
