```python
import csv
import os

# Stores employee details in a CSV file
class EmployeeFile:
    def __init__(self,file_name="employees.csv"):
        self.file_name=file_name
        # Make the file and add headings if it is not already there
        if not os.path.exists(self.file_name):
            with open(self.file_name,"w",newline="")as f:
                writer=csv.writer(f)
                writer.writerow(["emp_id","name","department","salary"])

    # Add a new employee
    def create(self,emp_id,name,department,salary):
        if self.find_employee(emp_id):
            print("Employee ID already exists")
            return
        with open(self.file_name,"a",newline="")as f:
            writer=csv.writer(f)
            writer.writerow([emp_id,name,department,salary])
        print("Employee added")

    # Display all employees
    def read_all(self):
        with open(self.file_name,"r",newline="")as f:
            rows=list(csv.reader(f))
        if len(rows)<=1:
            print("No employee records found")
            return
        for row in rows:
            print(row)

    # Change the details of an employee
    def update(self,emp_id,name,department,salary):
        with open(self.file_name,"r",newline="")as f:
            rows=list(csv.reader(f))
        found=False
        for row in rows[1:]:
            if row[0]==str(emp_id):
                row[1]=name
                row[2]=department
                row[3]=str(salary)
                found=True
                break
        if found:
            with open(self.file_name,"w",newline="")as f:
                writer=csv.writer(f)
                writer.writerows(rows)
            print("Employee updated")
        else:
            print("Employee not found")

    # Remove an employee using their ID
    def delete(self,emp_id):
        with open(self.file_name,"r",newline="")as f:
            rows=list(csv.reader(f))
        new_rows=[rows[0]]
        for row in rows[1:]:
            if row[0]!=str(emp_id):
                new_rows.append(row)
        if len(new_rows)==len(rows):
            print("Employee not found")
        else:
            with open(self.file_name,"w",newline="")as f:
                writer=csv.writer(f)
                writer.writerows(new_rows)
            print("Employee deleted")

    # Check whether an employee ID is already present
    def find_employee(self,emp_id):
        with open(self.file_name,"r",newline="")as f:
            reader=csv.reader(f)
            next(reader,None)
            for row in reader:
                if row and row[0]==str(emp_id):
                    return True
        return False

# Menu for using the CSV file
if __name__=="__main__":
    emp=EmployeeFile()
    while True:
        print("\n1. Create Employee")
        print("2. Read Employees")
        print("3. Update Employee")
        print("4. Delete Employee")
        print("5. Exit")
        choice=input("Enter choice: ")

        if choice=="1":
            emp.create(input("ID: "),input("Name: "),input("Department: "),input("Salary: "))
        elif choice=="2":
            emp.read_all()
        elif choice=="3":
            emp.update(input("ID: "),input("New name: "),input("New department: "),input("New salary: "))
        elif choice=="4":
            emp.delete(input("ID: "))
        elif choice=="5":
            break
        else:
            print("Invalid choice")
```
