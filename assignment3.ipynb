```python
import sqlite3

# Handles employee records in the database
class EmployeeDB:
    def __init__(self,db_name="company_a3.db"):
        self.db_name=db_name
        self.create_table()

    # Create the employee table if it does not exist
    def create_table(self):
        conn=sqlite3.connect(self.db_name)
        cur=conn.cursor()
        cur.execute("""CREATE TABLE IF NOT EXISTS employee (
            emp_id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            department TEXT NOT NULL,
            salary REAL NOT NULL
        )""")
        conn.commit()
        conn.close()

    # Insert a new employee record
    def create(self,emp_id,name,department,salary):
        conn=sqlite3.connect(self.db_name)
        cur=conn.cursor()
        try:
            cur.execute("INSERT INTO employee VALUES (?, ?, ?, ?)",(emp_id,name,department,salary))
            conn.commit()
            print("Employee added")
        except sqlite3.IntegrityError:
            print("Employee ID already exists")
        conn.close()

    # Show all employee records
    def read_all(self):
        conn=sqlite3.connect(self.db_name)
        cur=conn.cursor()
        cur.execute("SELECT * FROM employee")
        rows=cur.fetchall()
        conn.close()
        if not rows:
            print("No employee records found")
        else:
            for row in rows:
                print(row)

    # Update the record with the given ID
    def update(self,emp_id,name,department,salary):
        conn=sqlite3.connect(self.db_name)
        cur=conn.cursor()
        cur.execute("UPDATE employee SET name=?, department=?, salary=? WHERE emp_id=?",
                    (name,department,salary,emp_id))
        conn.commit()
        if cur.rowcount>0:
            print("Employee updated")
        else:
            print("Employee not found")
        conn.close()

    # Delete a record using employee ID
    def delete(self,emp_id):
        conn=sqlite3.connect(self.db_name)
        cur=conn.cursor()
        cur.execute("DELETE FROM employee WHERE emp_id=?",(emp_id,))
        conn.commit()
        if cur.rowcount>0:
            print("Employee deleted")
        else:
            print("Employee not found")
        conn.close()

# Menu for database operations
if __name__=="__main__":
    emp=EmployeeDB()
    while True:
        print("\n--- Database CRUD Menu ---")
        print("1. Create Employee")
        print("2. Read Employees")
        print("3. Update Employee")
        print("4. Delete Employee")
        print("5. Exit")
        choice=input("Enter choice: ")

        if choice=="1":
            try:
                emp.create(int(input("ID: ")),input("Name: "),input("Department: "),float(input("Salary: ")))
            except ValueError:
                print("Enter valid numbers")
        elif choice=="2":
            emp.read_all()
        elif choice=="3":
            try:
                emp.update(int(input("ID: ")),input("New name: "),input("New department: "),float(input("New salary: ")))
            except ValueError:
                print("Enter valid numbers")
        elif choice=="4":
            try:
                emp.delete(int(input("ID: ")))
            except ValueError:
                print("Enter a valid ID")
        elif choice=="5":
            break
        else:
            print("Invalid choice")
```
