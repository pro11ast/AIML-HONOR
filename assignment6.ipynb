```python
import time
import numpy as np
from numba import jit

# Prints how long a function takes to run

def profile_time(func):
    def wrapper(*args,**kwargs):
        start=time.perf_counter()
        result=func(*args,**kwargs)
        end=time.perf_counter()
        print(func.__name__,"time:",round((end-start)*1000,4),"ms")
        return result
    return wrapper

# Runs the simulation using normal Python
@profile_time
def normal_calculation(n):
    total=0.0
    for i in range(n):
        total=total+(i**0.5)*np.sin(i)
    return total

# Same calculation, but this function is compiled by Numba
@jit(nopython=True)
def fast_calculation(n):
    total=0.0
    for i in range(n):
        total=total+(i**0.5)*np.sin(i)
    return total

# Calls the Numba function and measures its time
@profile_time
def numba_calculation(n):
    return fast_calculation(n)

if __name__=="__main__":
    n=5000000
    print("Number of simulations:",n)

    print("\nNormal Python")
    normal_result=normal_calculation(n)

    print("\nNumba first run (includes compilation)")
    numba_result=numba_calculation(n)

    # Run it again after compilation to compare execution time
    print("\nNumba second run")
    numba_result=numba_calculation(n)
```
